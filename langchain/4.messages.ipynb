# Messages
In LangChain, Messages are the typed data structures that form the conversation history and state in an agentic loop. Instead of passing plain strings, LangChain uses specific message classes like HumanMessage, AIMessage, and ToolMessage. This explicit typing is critical for agents because it allows the LLM to distinguish between what the user asked, what the model previously decided to do, and the raw output returned by an external tool execution.


### What are Messages in LangChain? 
- Messages are the fundamental unit of context and communication for chat models in LangChain. They represent the input going into a model and the output coming back.Instead of just sending raw strings of text to an AI, LangChain uses structured Message objects to capture the state of a conversation, manage who is speaking, and handle complex data like tool usage or multimodal inputs (images, audio).
### The Anatomy of a MessageEvery message object in LangChain generally contains three core attributes:
- Role: Identifies who is sending the message (e.g., system, user, or assistant). This helps the chat model understand how to interpret and respond to the prompt.
- Content: The actual payload of the message. While this is most commonly text, it can also include multimodal data (like images or documents) or structured dictionaries.
- Metadata: Optional fields that carry extra information, such as unique message IDs, token usage counts, or specific tool call details.

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:qwen/qwen3.6-27b")
model.invoke("please tell what is AI")
```

## Text prompts
text prompt are stings - ideal for strightforward generation tasks where you don't need to retain conversation history

```python
model.invoke("what is langchain")
```

```python
from langchain.messages import SystemMessage,HumanMessage,AIMessage

messages = [
    SystemMessage("you are a pretry expert"),
    HumanMessage("write a poem on AI")
]

response = model.invoke(messages)
response
```

```python
system_msg = SystemMessage("you are a helpful coding assistant")

messages = [
    system_msg,
    HumanMessage("how do i create REST API")
]
response = model.invoke(messages)
print(response.content)
```

```python
## detailed info to LLM through system message

system_msg = SystemMessage("""
you are a senior python developer with expertise in web frameworks.
always provide code example and explain your resoning.
be concise but through in your explanations
""")

messages = [
    system_msg,
    HumanMessage("how do i create REST API")
]
response = model.invoke(messages)
print(response.content)
```

```python
human_msg = HumanMessage(
    content = "hello!",
    name = "alice",
    id = "msg_123"
)
response = model.invoke([human_msg])
response
```

```python
from langchain.messages import HumanMessage,SystemMessage,AIMessage

ai_msg = AIMessage("i'd be happy to help you with that question!")

message = [
    SystemMessage("you are helpful assistant"),
    HumanMessage("can you help me?"),
    ai_msg,
    HumanMessage("Great! what's 2+2")
]

response = model.invoke(message)
print(response.content)
```

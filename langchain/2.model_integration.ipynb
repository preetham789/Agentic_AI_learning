## GROQ model


```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:openai/gpt-oss-20b")
response = model.invoke("which one should focus on 2026 SDE or AI engineer in curreny job trend")
response
```

```python
import os
from langchain_groq import ChatGroq

model = ChatGroq(model = "openai/gpt-oss-20b")
response = model.invoke("which one should focus on 2026 SDE or AI engineer in curreny job trend")
response
```

```python
for chunk in model.stream("which one should focus on 2026 SDE or AI engineer in curreny job trend in paragraph"):
    print(chunk.text,end ="",flush = True)
```

## batch

```python
responses = model.batch([
    "why do fresher struggle to get job in IT sector in 2026"
    "how do get job faster"
    "what is AI engeneering"
])
for response in responses:
    print(response.content)
```

```python
responses = model.batch([
    "why do fresher struggle to get job in IT sector in 2026",
    "how do get job faster",
    "what is AI engeneering"],
    config = {
        'max_concurrency':5,
    }
)
for response in responses:
    print(response.content)
```
